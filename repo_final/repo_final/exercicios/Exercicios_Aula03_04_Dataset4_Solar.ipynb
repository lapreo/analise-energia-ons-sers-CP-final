## Situação

Uma equipe de operação de uma usina fotovoltaica deseja localizar períodos de alta geração e identificar quais inversores aparecem com maior frequência nesses momentos. O foco desta atividade é o arquivo de geração da planta (`Plant_1_Generation_Data.csv` ou `Plant_2_Generation_Data.csv`), sem junção com o arquivo de sensores meteorológicos.

## Etapa A — Orange Data Mining

Realizada no aplicativo **Orange Data Mining**, a partir do fluxo `FLUXO_BASE.ows` fornecido na disciplina. Os arquivos de resultado (fluxo `.ows` e amostra exportada em `.csv`) foram incluídos neste mesmo repositório, na pasta `orange/`.

Passos executados:

1. Abertura do arquivo de geração da planta (`Plant_1_Generation_Data.csv`) no widget **File**.
2. Inspeção dos atributos no **Data Table**.
3. Uso do **Select Columns** para manter `DATE_TIME`, `SOURCE_KEY`, `DC_POWER`, `AC_POWER`, `DAILY_YIELD` e `TOTAL_YIELD`.
4. Observação, no **Data Table**, da presença de registros com potência igual a zero (períodos noturnos/sem geração) e da faixa de valores de `AC_POWER`.
5. Geração de uma amostra aleatória de **20%** dos registros com o **Data Sampler**.
6. Exportação da amostra em CSV com o **Save Data**, salva como `amostra_solar_plant.csv`.

> O arquivo `amostra_solar_plant.csv` gerado nesta etapa é o ponto de partida da Etapa B, abaixo.

## Etapa B — Python / Pandas

Carregue o arquivo `amostra_solar_plant.csv` (exportado na Etapa A) na mesma pasta deste notebook, ou ajuste o caminho abaixo conforme necessário (por exemplo, `/content/amostra_solar_plant.csv` se estiver executando no Google Colab).

```python
import pandas as pd
from IPython.display import display

# Ajuste o caminho conforme onde o arquivo da amostra foi salvo
CAMINHO_AMOSTRA = "/content/amostra_solar_plant.tab"

df_solar = pd.read_csv(CAMINHO_AMOSTRA, sep="\t", skiprows=[1, 2])
```

**1. Inspeção inicial: `head()`, `shape`, `info()` e `describe()`**

```python
display(df_solar.head())
```

```python
print(f"Número de linhas: {df_solar.shape[0]}")
print(f"Número de colunas: {df_solar.shape[1]}")
```

```python
df_solar.info()
```

```python
df_solar.describe()
```

**2. Renomear as principais variáveis para nomes mais claros**

```python
df_solar = df_solar.rename(columns={
    "DC_POWER": "Potencia_CC",
    "AC_POWER": "Potencia_CA",
    "DAILY_YIELD": "Geracao_Diaria",
    "TOTAL_YIELD": "Geracao_Total"
})

display(df_solar.head())
```

**3. Determinar a maior potência CA registrada**

```python
potencia_ca_max = df_solar["Potencia_CA"].max()
print(f"Maior potência CA registrada na amostra: {potencia_ca_max:.2f} kW")
```

**4. Calcular o limiar correspondente a 70% da potência CA máxima**

```python
limiar_70 = 0.70 * potencia_ca_max
print(f"Limiar de alta geração (70% do máximo): {limiar_70:.2f} kW")
```

**5. Criar um DataFrame com registros acima do limite e calcular quantidade e percentual**

```python
df_alta_geracao = df_solar[df_solar["Potencia_CA"] > limiar_70]

qtd_alta_geracao = df_alta_geracao.shape[0]
pct_alta_geracao = qtd_alta_geracao / df_solar.shape[0] * 100

print(f"Registros em alta geração: {qtd_alta_geracao}")
print(f"Percentual em relação à amostra: {pct_alta_geracao:.2f}%")

display(df_alta_geracao.head())
```

**6. `value_counts()` em `SOURCE_KEY` dentro do DataFrame de alta geração**

```python
contagem_inversores = df_alta_geracao["SOURCE_KEY"].value_counts()
print(contagem_inversores)
```

**7. Inversor mais frequente nos registros de alta geração**

```python
inversor_mais_frequente = contagem_inversores.idxmax()
qtd_inversor_mais_frequente = contagem_inversores.max()

print(f"Inversor mais frequente em alta geração: {inversor_mais_frequente}")
print(f"Número de registros de alta geração desse inversor: {qtd_inversor_mais_frequente}")
```

**Interpretação:**

*(Complete após executar as células acima, com base nos valores obtidos.)* O inversor identificado como mais frequente em `contagem_inversores` aparece com mais medições acima do limiar de 70% da potência CA máxima do que os demais inversores da planta. Isso indica que, dentro do período e da amostra analisados, esse inversor esteve associado a mais momentos de alta geração — mas **não permite concluir, apenas com este recorte, se ele tem melhor desempenho ou se sofre menos perdas do que os outros**. Fatores como sombreamento localizado, orientação dos painéis conectados a cada inversor e eventuais falhas em outros inversores só poderiam ser avaliados com dados adicionais (por exemplo, o arquivo de sensores meteorológicos, que não foi utilizado nesta etapa).
